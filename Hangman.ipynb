```python
#keeping imports on top
import random

#10 word list
words = [
    "Potato",
    "Banana",
    "Peach",
    "Grape",
    "Broccoli",
    "Radish",
    "Pineapple",
    "Pear",
    "Lemon",
    "Olive",
    "Watermelon",
    "Strawberry",
    "Pomegranate",
    "Blackberry",
    "Cauliflower",
    "Artichoke",
    "Asparagus",
    "Cantaloupe",
    "Grapefruit",
    "Cranberry"
]

#Game state variables
game_state = {
    "GameOngoing": False,
    "Solution": "Potato",
    "Lives": 10,
    "GuessedLetters": [],
    "Display": ""
}
```

```python
#Creating each var independentley 

#GameOngoing

if game_state["GameOngoing"]:
    print("Game is on going")
else:
    print("Game over")

#Create a start and end func

def start_game():
    game_state["GameOngoing"] = True
    game_state["Solution"] = random.choice(words)
    game_state["Lives"] = 10
    game_state["GuessedLetters"] = []
    game_state["Display"] = ""

def end_game():
    game_state["GameOngoing"] = False
    game_state["Solution"] = ""
    game_state["Lives"] = 0
    game_state["GuessedLetters"] = []
    game_state["Display"] = ""
```

```python
#Sloution

game_state["Solution"] = random.choice(words)

print(game_state["Solution"])
print("The solution has ", len(game_state["Solution"]), "letters")

```

```python
#GuessedLetters

print(game_state["Solution"])
print("Current amount of lives: ", game_state["Lives"])

#Letter input and conditions
Letter = input("Input letter").lower()

if Letter in game_state["GuessedLetters"]:
    print(f"You already guessed the letter {Letter}")
else:
    game_state["GuessedLetters"].append(Letter)
    if Letter in game_state["Solution"].lower():
           print(f"Correct guess the letter {Letter.upper()} is in this word")
    else:
           game_state["Lives"]-= 1
           print(f"The letter {Letter.upper()} is not in this word")


print(game_state["GuessedLetters"])

print("The solution has ", len(game_state["Solution"]), "letters")
print("Current amount of lives: ", game_state["Lives"])
```

```python
#Display

def update_display():
    game_state["Display"] = ""

    for letter in game_state["Solution"]:
        if letter.lower() in game_state["GuessedLetters"]:
            game_state["Display"] += letter
        else:
            game_state["Display"] += "_"


print(game_state["Display"])
print(game_state["GuessedLetters"])
print("The solution has ", len(game_state["Solution"]), "letters")
print("Current amount of lives: ", game_state["Lives"])
```

```python
#Game Loop
start_game()

while game_state["GameOngoing"]:
    
    update_display()
    print(game_state["Display"])
    print(game_state["GuessedLetters"])
    print("The solution has ", len(game_state["Solution"]), "letters")
    print("Current amount of lives: ", game_state["Lives"])
    print(Hangman_pics[10 - game_state["Lives"]])

    Letter = input("Input letter").lower()
    

    if Letter in game_state["GuessedLetters"]:
        print(f"You already guessed the letter {Letter}")
    else:
        game_state["GuessedLetters"].append(Letter)
        if Letter in game_state["Solution"].lower():
            print(f"Correct guess the letter {Letter.upper()} is in this word")
        else:
            game_state["Lives"]-= 1
            print(f"The letter {Letter.upper()} is not in this word")

    update_display()
     

    # WIN
    if  game_state["Display"].lower() == game_state["Solution"].lower():
        print("\nYou win!")
        print("The word was:", game_state["Solution"])
        game_state["GameOngoing"] = False

    # LOSE
    elif game_state["Lives"] <= 0:
        print("\nGame over!")
        print(Hangman_pics[10])
        print("The word was:", game_state["Solution"])
        game_state["GameOngoing"] = False

# Reset everything
end_game()

print("\nGame has been reset.")
```
